**dividing our data into train_data and test_data** 

```python
holdout_days = 31

series = df_beverages['units_sold'].to_numpy(dtype=float)
dates = df_beverages['date'].to_numpy()

train_beverages = series[:-holdout_days]
test_beverages = series[-holdout_days:]
test_dates = dates[-holdout_days:]
```

```python
test_beverages
```

**predicting sales for the next 31 days using the holt winter forecast**

```python
def holt_winter_forecast(y,horizon,season_length=7):
    model=ExponentialSmoothing(
        y,
        trend='add',
        seasonal='add',
        seasonal_periods=season_length
    )
    fitted_model = model.fit()
    forecast = fitted_model.forecast(horizon)
    return np.maximum(0,forecast)

Sales = train_beverages
future = holt_winter_forecast(Sales,holdout_days)
print(future)


#basically takes the last 84 days of the train data, observes trends, seasonality,and use those observations to predict values for the next 31 days
```

```python
forecast_table= pd.DataFrame({'Forecast':future,
                                 'Actual':test_beverages})
print(forecast_table)
```

```python
forecast_table['AbsoluteError'] = (forecast_table['Actual']-forecast_table['Forecast']).abs()
```

```python
forecast_table
```

```python
mae= forecast_table['AbsoluteError'].mean()
print(mae)
```

```python
rmse = np.sqrt(((forecast_table['Actual']-forecast_table['Forecast'])**2).mean())
print(rmse)
```

### Seasonal Naive

```python
df_beverages
```

```python
holdout_days = 31
season_length = 7

series = df_beverages['units_sold'].to_numpy(dtype=float)
dates = df_beverages['date'].to_numpy()

if len(series)< (holdout_days + season_length):
    raise ValueError('Not enough history for a 31 day hold out and a 7 day season!')

train_data = series[:-holdout_days]
test_data = series[-holdout_days:]
test_dates=dates[-holdout_days:]

print(f"Training data: {len(train_data)} days")
print(f"Holdout/testing data: {len(test_data)} days ")

def seasonal_naive_forecast(y, horizon, season=7):
    last_season = y[-season:]
    reps = int(np.ceil(horizon/season))
    x = np.tile(last_season,reps)[:horizon]
    return x


future = seasonal_naive_forecast(train_data,holdout_days)
print(future)
    
```

### Moving 7 day average

```python
df_beverages
```

```python
holdout_days = 31
season_length = 7

series = df_beverages['units_sold'].to_numpy(dtype=float)
dates = df_beverages['date'].to_numpy()

if len(series) < holdout_days + season_length:
    raise ValueError("Not enough history for a 31-day holdout plus a 7-day window.")
    
train_data = series[:-holdout_days]
test_data = series[-holdout_days:]
test_dates=dates[-holdout_days:]

print(f"Training data: first {len(train_data)} days of our dataset")
print(f"Testing_data: the last {len(test_data)} days of our dataset")

def moving_average(y,steps, season_length=7):
    history = list(y)
    future =[]

    for _ in range(steps):
        avg = np.mean(history[-season_length:])
        future.append(avg)
        history.append(avg)
    x = np.array(future)
    return x

future_x = moving_average(train_data,holdout_days)
print(future_x)
    
    
 
    
```

```python
Actual = df_beverages['units_sold'].iloc[-31:]
```

```python
forecast_table = pd.DataFrame({'Actual':Actual,
                               'Forecast':future_x})
forecast_table
```

```python
forecast_table['AbsoluteError']= (forecast_table['Actual']-forecast_table['Forecast']).abs()
forecast_table
```

```python
mae = forecast_table['AbsoluteError'].mean()
print(mae)
```

```python
rmse = np.sqrt(((forecast_table['Actual']-forecast_table['Forecast'])**2).mean())
print(rmse)
```

The above method for the 7 day moving average was recursive .We decided to use the average of the last 7 days of our training data to forcast the next value, then we kept on moving the window one value to the next for the rest of the 31 days hold out.
Note this method is not so accurate since it evens out the trends and seasonality so it pretty much doesnt capture it. The best method is to use the last 7 days mean to forecast the next 31 days

```python
def rolling_average(y, steps, season_length=7):
    avg = y[-season_length:].mean()
    x =np.repeat(avg, steps)
    return x

future = rolling_average(train_data, holdout_days) 
print(future)

#currently we are observing from the last 7 days of our data 
```

```python
Actual = df_beverages['units_sold'].iloc[-31:]
```

```python
forecast_table = pd.DataFrame({'Actual':Actual,
                               'Forecast':future})
forecast_table
```

```python
forecast_table['AbsoluteError']= (forecast_table['Actual']-forecast_table['Forecast']).abs()
forecast_table
```

```python
mae = forecast_table['AbsoluteError'].mean()
print(mae)
```

```python
rmse = np.sqrt(((forecast_table['Actual']-forecast_table['Forecast'])**2).mean())
print(rmse)
```

Avoiding the recursive method, will generate more acurate values.You can see that from the **rmse** and **mae**. 

**NOTE:** currently we are observing our data from a window of 7 days, taking the average of those last 7 days from our training data to predict the next 31 values and find errors.At 7(`mae`:7.2, `rmse`:9.8).At 28(`mae`:7.3, `rmse`:9.9).At 84 (`mae`:7.2, `rmse`:9.9).

```python
#SUMMARY
#Moving 7 day average forecast, window of 7 days : mae: 7.2, rmse: 9.8
#Holt winter(ExponentialSmoothing) forecast, window of 84 days: mae:5.6, rmse:7.3
#Holt winter forecast, window of 7 days: mae:6.3, rmse:8.0
```

We can now decide to predict values for our product using Holt winter (ExponentialSmoothing).The method shows more accuracy than the rest.We will predict values for the next 14 days for that specific product , and we will use a window of 7 days.

```python

series = df_beverages['units_sold'].to_numpy(dtype=float)
date = df_beverages['date'].to_numpy()
series

train_data = series

def holt_winter(y, horizon, seasonal_periods=7):
    model=ExponentialSmoothing(
        y,
        trend='add',
        seasonal='add',
        seasonal_periods=seasonal_periods
     )
    fitted_model = model.fit()
    forecast = fitted_model.forecast(horizon)
    return np.maximum(0,forecast)

future = holt_winter(train_data,14)
x=(np.round(future,0))
print(x)
```
